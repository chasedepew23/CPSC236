## 3.1 Letter Grade Converter
Create a program that converts number grades to letter grades.

### Console:
```powershell
Letter Grade Converter

Enter numerical grade: 90 
Letter grade: A

Continue? (y/n): y

Enter numerical grade: 88 
Letter grade: A

Continue? (y/n): y

Enter numerical grade: 80 
Letter grade: B

Continue? (y/n): y

Enter numerical grade: 67 
Letter grade: C

Continue? (y/n): y

Enter numerical grade: 59 
Letter grade: F

Continue? (y/n): n

Bye!
```

### Specifications:
- Assume that the user will enter valid integers for the grades.
- The program should continue only if the user enters “y” or “Y” to continue.
- The grading criteria is as follows:

| Letter |	Range |
|---|-------|
| A |	88-100 |
| B |	80-87 |
| C |	67-79 |
| D |	60-66 |
| F |	<60 |

````python
### CODE HERE ###
#Chase Depew: CPSC 236 3.1 Letter Grade Convertor

print("Letter Grade Convertor")

choice = "y"
while choice.lower() == "y" or choice.lower == "Y":
    grade = int(input("Enter Numerical Grade: "))
    if grade >= 88:
        print("Letter Grade: A")
    if grade >= 80 and grade < 88:
        print("Letter Grade: B")
    if grade >= 67 and grade < 80:
        print("Letter Grade: C")
    if grade >= 60 and grade < 67:
        print("Letter Grade: D")
    if grade < 60:
        print("Letter Grade: F") 
    choice = input("Continue? (y/n): ")
print("Bye!")
````

### Specifications:

````python
### CODE HERE ###
#Chase Depew CPSC 236 3.2 Tip Calculator

print("Tip Calculator\n")

cost = float(input("Cost of meal: "))

print("15%")
tipamount15 = cost * 0.15
print("Tip amount: ", float(round(tipamount15, 2)))
totalamount15 = cost + tipamount15
print("Total amount: ", float(round(totalamount15, 2))) 

print("20%")
tipamount20 = cost * 0.20
print("Tip amount: ", float(round(tipamount20, 2)))
totalamount20 = cost + tipamount20
print("Total amount: ", float(round(totalamount20, 2)))

print("25%")
tipamount25 = cost * 0.25
print("Tip amount: ", float(round(tipamount25, 2)))
totalamount25 = cost + tipamount25
print("Total amount: ", float(round(totalamount25, 2)))
````

## 3.3 - Change Calculator
Create a program that calculates the coins needed to make change for the specified number of cents.

### Console
```powershell
Change Calculator

Enter number of cents (0-99): 99

Quarters: 3
Dimes:	2
Nickels: 0
Pennies: 4

Continue? (y/n): y

Enter number of cents (0-99): 55

Quarters: 2
Dimes:	0
Nickels: 1
Pennies: 0

Continue? (y/n): n 

Bye!
```

### Specifications
- The program should display the minimum number of quarters, dimes, nickels, and pennies that one needs to make up the specified number of cents.
- Assume that the user will enter a valid integer for the number of cents.
- The program should continue only if the user enters “y” or “Y” to continue.


````python
### CODE HERE ###
### CPSC 236 Chase Depew 3.3 Change Calculator

print("Change Calculator")

while True:
    cents = int(input("\nEnter number of cents (0-99): "))

    quarters = cents // 25
    cents = cents % 25

    dimes = cents // 10
    cents = cents % 10

    nickels = cents // 5
    cents = cents % 5

    pennies = cents

    print(f"\nQuarters: {quarters}")
    print(f"Dimes:\t{dimes}")
    print(f"Nickels: {nickels}")
    print(f"Pennies: {pennies}")

    choice = input("\nContinue? (y/n): ")

    if choice.lower() != "y":
        print("\nBye!")
        break
````

## 3.4 - Table of Powers
Create a program that displays a table of squares and cubes for the specified range of numbers.

### Console
```powershell
Table of Powers

Start number: 90
Stop number: 100

Number	Squared	Cubed
======	=======	=====
90      8100	729000
91      8281	753571
92      8464	778688
93      8649	804357
94      8836	830584
95      9025	857375
96      9216	884736
97      9409	912673
98      9604	941192
99      9801	970299
100     10000	1000000
```

### Specifications
- The formulas for calculating squares and cubes are:

    `square = x ** 2`
    
    `cube = x ** 3`

- Use tabs to align the columns.
- Assume that the user will enter valid integers.
- Make sure the user enters a start integer that’s less than the stop integer. If the user enters a start integer that’s greater than the stop integer, display an error message and give the user a chance to enter the integers again.


````python
### CODE HERE ###
### CPSC 236 Chase Depew 3.4 Table of Powers

print("Table of Powers")

while True:
    start = int(input("Start number: "))
    stop = int(input("Stop number: "))

    if start < stop:
        break
    else:
        print("Error: Start number must be less than stop number. Please try again.\n")

print("\nNumber\tSquared\tCubed")
print("======\t=======\t=====")

for x in range(start, stop + 1):
    square = x ** 2
    cube = x ** 3
    print(f"{x}\t{square}\t{cube}")
````
