```python
### Chase Depew 2/25/26 4.1 Even or Odd Checker ###

def check_even_or_odd(number):
    if number % 2 == 0:
        return "even"
    else:
        return "odd"


def main():
    print("Even or Odd Checker")
    
    number = int(input("Enter an integer: "))
    
    result = check_even_or_odd(number)
    
    print(f"This is an {result} number.")


if __name__ == "__main__":
    main()
```

```python
### Chase Depew 2/25/26 4.2 Feet and Meters Convertor with seperate module file ###

import sys
import conversions


# Now the main program
def display_title():
    print("Feet and Meters Converter\n")

def display_menu():
    print("Conversions Menu:")
    print("a. Feet to Meters")
    print("b. Meters to Feet")

def main():
    display_title()
    
    again = "y"
    
    while again.lower() == "y":
        display_menu()
        choice = input("Select a conversion (a/b): ").lower()
        
        if choice == "a":
            feet = float(input("\nEnter feet: "))
            meters = conversions.feet_to_meters(feet)
            print(f"{round(meters, 2)} meters")
        
        elif choice == "b":
            meters = float(input("\nEnter meters: "))
            feet = conversions.meters_to_feet(meters)
            print(f"{round(feet, 2)} feet")
        
        again = input("\nWould you like to perform another conversion? (y/n): ")
        print()
    
    print("Thanks, bye!")

if __name__ == "__main__":
    main()
```

```python
### Chase Depew 2/25/26 4.3 Sales Tax Calculator with seperate module file ###

import sales_tax


def get_total():
    ### Prompt user to enter item costs and return total.###
    total = 0.0
    print("ENTER ITEMS (ENTER 0 TO END)")
    
    while True:
        cost = float(input("Cost of item: "))
        if cost == 0:
            break
        total += cost
    
    return round(total, 2)


def display_totals(total):
    ### Display total, sales tax, and total after tax.###
    tax = sales_tax.calculate_tax(total)
    total_after_tax = sales_tax.calculate_total_after_tax(total)

    print(f"Total:  {total}")
    print(f"Sales tax:\t{tax}")
    print(f"Total after tax: {total_after_tax}")
    print()


def main():
    print("Sales Tax Calculator\n")

    while True:
        total = get_total()
        display_totals(total)

        again = input("Again? (y/n): ").lower()
        print()
        if again != 'y':
            break

    print("Thanks, bye!")


if __name__ == "__main__":
    main()
```

```python
### Chase Depew 2/25/26 4.4 Prime Number Checker

def get_valid_integer():
    ### Prompt user for an integer between 1 and 5000.###
    while True:
        num = int(input("Please enter an integer between 1 and 5000: "))
        if 1 < num <= 5000:
            return num
        else:
            print("Invalid integer. Please try again.")


def count_factors(number):
    ### Return the total number of factors of a number.###
    factor_count = 0
    for i in range(1, number + 1):
        if number % i == 0:
            factor_count += 1
    return factor_count


def main():
    print("Prime Number Checker\n")

    while True:
        number = get_valid_integer()
        factors = count_factors(number)

        if factors == 2:
            print(f"{number} is a prime number.\n")
        else:
            print(f"{number} is NOT a prime number.")
            print(f"It has {factors} factors.\n")

        again = input("Try again? (y/n): ").lower()
        print()
        if again != 'y':
            break

    print("Bye!")


if __name__ == "__main__":
    main()
```
